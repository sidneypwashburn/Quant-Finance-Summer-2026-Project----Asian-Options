# Asian Option Pricing by the Večeř Partial Differential Equation

## 1. Introduction

In this notebook, the price of a fixed-strike Asian option is computed under the Black–Scholes model using the one-dimensional Večeř PDE. Our main reference is *Stochastic Calculus for Finance II: Continuous-Time Models* by Steven E. Shreve.

We beging with a risk-neutral model
$$
dS(t) = r S(t)\,dt + \sigma S(t)\,dW(t).
$$
Here:
- $S(t)$: Stock price at time $t$,
- $r$: Continuously compounded risk-free rate,
- $\sigma$: Volatility,
- $T$: Maturity.

Consider a continuously sampled fixed-strike Asian call option with payoff at maturity
$$
V(T)=\left(\frac{1}{T}\int_0^T S(t)\,dt - K\right)^+.
$$
Define now for $0\leq t\leq T$ the auxiliary process
$$
Z(t)=q(t)+\frac{e^{-r(T-t)}}{S(t)}\left(\frac{1}{T}\int_0^t S(u)\,du-K\right).
$$
Here $q(t)$ is given as
$$
q(t)=
\begin{cases}
    \dfrac{1-e^{-r(T-t)}}{rT} & r\neq 0\\[1 mm]
    \dfrac{T-t}{T} & r=0
\end{cases}.
$$
After changing the numéraire $Z(t)$ represents the value of the portfolio expressed in units of the risky asset rather than in dollars.

Now the call price at any time $0\leq t\leq T$ can be written as
$$
V(t)=S(t)\,u(t,Z(t)).
$$
Here $u(t,z)$ satisfies the following PDE
$$
u_{t}(t,z)+\frac{1}{2}\sigma^{2}(q(t)-z)^{2}u_{zz}(t,z)=0,
$$
with terminal condition
$$
u(T,z)=z^{+}.
$$
In the case of a put option the same PDE applies, with the terminal condition changed to
$$
u(T,z)=(-z)^{+}.
$$
Other than this, the analysis remains the same.

We introduce the time-to-maturity variable $\tau=T-t$. Defining $v(\tau,z)=u(T-\tau,z)$ transforms the backward PDE into the forward-time equation
$$
v_\tau(\tau,z)=\frac{1}{2}\sigma^2(q(T-\tau)-z)^2v_{zz}(\tau,z).
$$
We will apply numerical methods to solve this forward-time equation.

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.sparse import diags
from scipy.sparse.linalg import spsolve
```

## 2. Numerical Solution 

Our objective is to solve the PDE
$$
v_\tau(\tau,z)=\frac{1}{2}\sigma^2(q(T-\tau)-z)^2v_{zz}(\tau,z).
$$
In order to simplify the notation, let us write
$$
a(\tau,z)=\frac12\sigma^2(q(\tau)-z)^2.
$$
An implicit Euler finite-difference scheme is used to discretize the equation. Denote as $\Delta z$ and $\Delta\tau$ the spatial and temporal discretization steps, respectively. Define next the grid points by
$$
z_{j}=z_{\min}+j\Delta z\,,\qquad \tau_{i}=i\Delta\tau\,,\qquad i,j\geq 0.
$$

We approximate the time derivative by the forward difference
$$
v_{\tau}(\tau_{i},z)
\approx
\frac{v(\tau_{i+1},z)-v(\tau_{i},z)}{\Delta\tau}.
$$
Similarly, the second-order spatial derivative is approximated using the central difference formula
$$
v_{zz}(\tau,z_j)
\approx
\frac{v(\tau,z_{j+1})-2v(\tau,z_j)+v(\tau,z_{j-1})}{(\Delta z)^2}.
$$
Combining these approximations yields a tridiagonal linear system at each time step
$$
\frac{v(\tau_{i+1},z_{j})-v(\tau_{i},z_{j})}{\Delta\tau}=
a(\tau_{i+1},z_{j})\frac{v(\tau_{i+1},z_{j+1})-2v(\tau_{i+1},z_j)+v(\tau_{i+1},z_{j-1})}{(\Delta z)^2}
$$

To simplify notation, let us write
$$
v_{i,j}=v(\tau_i,z_j),\qquad\lambda_{i+1,j}=\frac{\Delta\tau}{(\Delta z)^2}a(\tau_{i+1},z_j).
$$
Rearranging the equation above gives
$$
-\lambda_{i+1,j}v_{i+1,j-1}+(1+2\lambda_{i+1,j})v_{i+1,j}-\lambda_{i+1,j}v_{i+1,j+1}=v_{i,j}.
$$
In any fixed time step from $\tau_i$ to $\tau_{i+1}$, the unknown values $v_{i+1,j}$ form the unkown vector
$$
\mathbf{v}_{i+1}=(v_{i+1,1},v_{i+1,2},\ldots,v_{i+1,N})^{t}.
$$
We can rewrite the finite-difference equations in matrix form as
$$
A_{i+1}\mathbf{v}_{i+1}=\mathbf{v}_{i}
$$
where $A_{i+1}$ is the tridiagonal matrix
$$
A_{i+1}=
\begin{pmatrix}
1+2\lambda_{i+1,1} & -\lambda_{i+1,1} & 0 & \cdots & 0\\
-\lambda_{i+1,2} & 1+2\lambda_{i+1,2} & -\lambda_{i+1,2} & \ddots & \vdots\\
0 & -\lambda_{i+1,3} & 1+2\lambda_{i+1,3} & \ddots & 0\\
\vdots & \ddots & \ddots & \ddots & -\lambda_{i+1,N-1}\\
0 & \cdots & 0 & -\lambda_{i+1,N} & 1+2\lambda_{i+1,N}
\end{pmatrix}.
$$
Solving this linear system yields the unknown vector, which provides the solution at the next time level. Repeating this process allows the numerical solution to be advanced iteratively through all time steps.

## 3 . Implementation

Pricing fixed-strike Asian options requires the following financial inputs:

- `S0`: Initial stock price $S(0)$
- `K`: Strike price $K$
- `r`: Continuously compounded risk-free rate
- `sigma`: Volatility
- `T`: Maturity in years
- `option_type`: `"call"` or `"put"`

On the other hand, the numerical inputs controlling the finite-difference discretization are:

- `z_min`, `z_max`: Lower and upper bounds of the $z$-grid
- `Nz`: Number of grid points in the $z$-direction
- `Nt`: Number of time steps

```python
# Financial inputs
S0 = 100.0
K = 100.0
r = 0.05
sigma = 0.20
T = 1.0

# Option type: choose "call" or "put"
option_type = "call"

# Numerical grid inputs
z_min = -1.5
z_max = 1.5
Nz = 801
Nt = 1000
```

In the transformed Večeř PDE the function $q(t)$ is given as
$$
q(t)=
\begin{cases}
    \dfrac{1-e^{-r(T-t)}}{rT} & r\neq 0\\[1 mm]
    \dfrac{T-t}{T} & r=0
\end{cases}.
$$
Since the implementation uses the time-to-maturity variable ($\tau=T-t$), the equivalent expression is
$$
q(\tau)=
\begin{cases}
    \dfrac{1-e^{-r\tau}}{rT} & r\neq 0\\[1 mm]
    \dfrac{\tau}{T} & r=0
\end{cases}.
$$

```python
def q_continuous_tau(tau, r, T):
    """
    Computes q as a function of tau = T - t.

    Parameters
    ----------
    tau : float
        Forward time variable tau = T - t.
    r : float
        Risk-free interest rate.
    T : float
        Maturity.

    Returns
    -------
    float
        The value q(tau) = (1 - exp(-r tau)) / (r T),
        with the correct limiting formula when r is close to zero.
    """
    if abs(r) < 1e-12:
        return tau / T
    return (1.0 - np.exp(-r * tau)) / (r * T)
```

With these definitions in place, we can now implement the finite-difference method described above to solve Večeř's PDE and compute the price of a fixed-strike Asian option. 

By default, the function returns only the option price. If `return_grid=True`, it also returns:
- `z0`: the transformed initial state,
- `z`: the numerical $z$-grid,
- `v`: the computed PDE solution at $\tau=T$.

```python
def price_asian_vecer_fixed_strike(
    S0,
    K,
    r,
    sigma,
    T,
    option_type="call",
    z_min=-1.5,
    z_max=1.5,
    Nz=801,
    Nt=1000,
    return_grid=False
):
    """
    Prices a continuously sampled fixed-strike Asian option using the one-dimensional Věčer PDE.

    Parameters
    ----------
    S0 : float
        Initial stock price.
    K : float
        Strike price.
    r : float
        Continuously compounded risk-free rate.
    sigma : float
        Volatility.
    T : float
        Maturity in years.
    option_type : str
        Either "call" or "put".
    z_min : float
        Lower bound of the spatial z-grid.
    z_max : float
        Upper bound of the spatial z-grid.
    Nz : int
        Number of z-grid points.
    Nt : int
        Number of time steps.
    return_grid : bool
        If True, return the full z-grid and final PDE solution.

    Returns
    -------
    price : float
        The Asian option price.
    z0 : float
        The transformed initial state.
    z : numpy.ndarray
        The z-grid, returned only if return_grid=True.
    v : numpy.ndarray
        The PDE solution at tau=T, returned only if return_grid=True.
    """

    # ------------------------------------------------------------
    # 1. Build the spatial grid in z.
    # ------------------------------------------------------------
    z = np.linspace(z_min, z_max, Nz)
    dz = z[1] - z[0]
    dt = T / Nt

    # ------------------------------------------------------------
    # 2. Initial condition in tau.
    # ------------------------------------------------------------
    option_type = option_type.lower()

    if option_type == "call":
        v = np.maximum(z, 0.0)
    elif option_type == "put":
        v = np.maximum(-z, 0.0)
    else:
        raise ValueError("option_type must be either 'call' or 'put'.")

    # ------------------------------------------------------------
    # 3. Boundary conditions.
    #
    # For a call:
    #   as z -> -infinity, v -> 0
    #   as z -> +infinity, v ~ z
    #
    # For a put:
    #   as z -> -infinity, v ~ -z
    #   as z -> +infinity, v -> 0
    # ------------------------------------------------------------
    if option_type == "call":
        left_boundary = 0.0
        right_boundary = z_max
    else:
        left_boundary = -z_min
        right_boundary = 0.0

    # ------------------------------------------------------------
    # 4. Time stepping.
    #
    # At each step we solve:
    #
    #   v_new - dt * a * Dzz(v_new) = v_old
    # ------------------------------------------------------------
    for n in range(Nt):
        tau_next = (n + 1) * dt

        q = q_continuous_tau(tau_next, r, T)

        # Interior grid points only.
        z_interior = z[1:-1]

        # Diffusion coefficient a(tau,z).
        a = 0.5 * sigma**2 * (q - z_interior)**2

        # lambda_j = dt * a_j / dz^2
        lam = dt * a / dz**2

        # Tridiagonal matrix:
        #
        # -lam_j * v_{j-1}^{new}
        # + (1 + 2 lam_j) * v_j^{new}
        # - lam_j * v_{j+1}^{new}
        # = v_j^{old}
        lower_diagonal = -lam[1:]
        main_diagonal = 1.0 + 2.0 * lam
        upper_diagonal = -lam[:-1]

        A = diags(
            diagonals=[lower_diagonal, main_diagonal, upper_diagonal],
            offsets=[-1, 0, 1],
            format="csc"
        )

        rhs = v[1:-1].copy()

        # Add boundary-condition contributions.
        rhs[0] += lam[0] * left_boundary
        rhs[-1] += lam[-1] * right_boundary

        # Solve for the new interior values.
        v_new_interior = spsolve(A, rhs)

        # Reconstruct the full solution including boundaries.
        v[0] = left_boundary
        v[-1] = right_boundary
        v[1:-1] = v_new_interior

    # ------------------------------------------------------------
    # 5. Compute the initial transformed state z0.
    #
    # At t=0 it holds that
    #
    # z0 = q(0) - exp(-rT) K/S0.
    #
    # In tau variables, q(0) means q_tau(T).
    # ------------------------------------------------------------
    q0 = q_continuous_tau(T, r, T)
    z0 = q0 - np.exp(-r * T) * K / S0

    # Interpolate the PDE solution at z0.
    v_at_z0 = np.interp(z0, z, v)

    # Recover the actual option price.
    price = S0 * v_at_z0

    if return_grid:
        return price, z0, z, v

    return price
```

```python
# ------------------------------------------------------------
#  Testing
# ------------------------------------------------------------

price, z0, z_grid, v_grid = price_asian_vecer_fixed_strike(
    S0=S0,
    K=K,
    r=r,
    sigma=sigma,
    T=T,
    option_type=option_type,
    z_min=z_min,
    z_max=z_max,
    Nz=Nz,
    Nt=Nt,
    return_grid=True
)

print(f"Option type: {option_type}")
print(f"Asian option price: {price:.6f}")
print(f"Initial transformed state z0: {z0:.6f}")
```

We can plot the PDE solution as a function of the transformed state variable. Below the horizontal axis corresponds to $z$, and the vertical axis corresponds to the PDE solution $v(T,z)$.

Note that the actual option price is obtained by evaluating this function at $z=z_0$ and multiplying the result by the initial stock price $S_0$.

```python
plt.figure(figsize=(8, 5))
plt.plot(z_grid, v_grid, label="PDE solution v(T,z)")
plt.axvline(z0, linestyle="--", label=f"z0 = {z0:.4f}")
plt.xlabel("z")
plt.ylabel("v(T,z)")
plt.title("Věčer PDE solution for fixed-strike Asian option")
plt.legend()
plt.grid(True)
plt.show()
```
